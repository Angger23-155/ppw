# Pengantar Web Mining

Secara sederhana, **Web Mining** adalah penerapan teknik *data mining* untuk menemukan pola, informasi, atau pengetahuan yang bermanfaat dari data di World Wide Web (WWW). 

Berbeda dengan data tradisional yang terstruktur (seperti *database* relasional), data di web sangat besar, tersebar, dan sebagian besar tidak terstruktur (teks HTML, gambar, tautan). Oleh karena itu, Web Mining dikelompokkan menjadi tiga kategori utama:

**1. Web Content Mining**
* Fokus pada proses penggalian informasi yang berguna dari isi atau konten suatu halaman web.
* **Sumber Data:** Teks, gambar, audio, video, atau metadata dalam halaman HTML.
* **Contoh Penerapan:** Analisis sentimen pada ulasan produk, klasifikasi topik artikel berita, atau penyaringan spam.

**2. Web Structure Mining**
* Fokus pada analisis struktur tautan (*hyperlink*) antar halaman web untuk menemukan topologi atau arsitektur dari sebuah situs.
* **Sumber Data:** Struktur graf web (Node = halaman web, Edge = hyperlink antar halaman).
* **Contoh Penerapan:** Algoritma *PageRank* milik Google untuk menentukan seberapa penting atau otoritatif suatu halaman web berdasarkan jumlah *link* yang mengarah kepadanya.

**3. Web Usage Mining**
* Fokus pada analisis pola interaksi dan perilaku pengguna saat mereka berselancar di sebuah situs web.
* **Sumber Data:** *Log server* (catatan aktivitas), *clickstream* (alur klik), dan riwayat durasi kunjungan pengguna.
* **Contoh Penerapan:** Sistem rekomendasi produk di e-commerce ("*Pengguna yang melihat barang ini juga membeli...*"), atau analisis untuk memperbaiki desain tata letak (UI/UX) website.

**Kesimpulan Singkat**
Jika *Content Mining* mencari tahu **apa isi webnya**, *Structure Mining* memetakan **bagaimana web itu saling terhubung**, maka *Usage Mining* mempelajari **bagaimana orang menggunakan web tersebut**.
